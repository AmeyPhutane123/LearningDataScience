### 📌 Topic 1: Introduction to Forecasting & Time Series Analysis

### Definition

**Forecasting** is the process of predicting **future values** based on **historical data**.

In **time series forecasting**, data is collected at **regular time intervals**  
(daily, monthly, yearly, etc.), and the goal is to **project future values** by  
analyzing **past patterns**.

**Example:**  
Predicting next month’s electricity consumption based on monthly usage for the past 3 years.

### 📌 Concept

A **time series** is an **ordered sequence** of data points recorded over time.

When we forecast, we assume that:

- **Past patterns** can help us understand the future.
- Data may contain **trend**, **seasonality**, and **random fluctuations**.



**Common components:**
- **Trend** – long-term increase or decrease
- **Seasonality** – repeating patterns at fixed intervals
- **Cyclic** – repeating patterns without fixed frequency
- **Irregular (Noise)** – random, unpredictable variation

- Some methods rely purely on **past data** (statistical),
  while others incorporate **external factors** (machine learning).


### 📌 Basic Math Example — Naive Forecasting

One of the simplest forecasting methods is the **naive method**, where:

$$
\text{Forecast for next period} = \text{Value of last period}
$$

- This method works surprisingly well for short- term and stable series but fails when trends or seasonality are strong.

### 📌 Components of a Time Series

### Definition

A time series is made up of several **underlying patterns** that, when combined,  
produce the **observed data**. Breaking the series into its components helps us  
understand its behavior and select the **right forecasting method**.

### Concept

There are **four main components** in a time series:

- **Level** – The baseline average value of the series, around which variations occur.

- **Trend** – The long-term upward or downward movement.  
  **Example:** Steady sales growth over years.

- **Seasonality** – Regular, repeating patterns at fixed intervals.  
  **Example:** Ice cream sales peaking every summer.

- **Noise (Irregular)** – Random fluctuations caused by unpredictable factors,  such as sudden events or errors in data.

### 📌 Time Series Models

The combination of these components can be **additive** or **multiplicative**.

#### ➕ Additive Model

$$
Y_t = \text{Level} + \text{Trend} + \text{Seasonality} + \text{Noise}
$$

Used when seasonal variations are **roughly constant over time**.

---

#### ✖️ Multiplicative Model

$$
Y_t = \text{Level} \times \text{Trend} \times \text{Seasonality} \times \text{Noise}
$$

Used when seasonal variations **increase or decrease with the trend**.

# Data Preparation Techniques

### Definition

Data preparation in time series involves modifying, cleaning, or restructuring the dataset to make it suitable for analysis and forecasting.

### Concept

In time series, raw data often needs:

Resampling: Changing the frequency of data points.

Interpolation: Filling missing values when frequency is increased or data is incomplete.

Transformation: Applying mathematical functions to stabilize variance or make the series more linear.

### Part 1 — Resampling

Why

Sometimes data is recorded at the wrong frequency for your analysis (e.g., daily when you need monthly, or monthly when you need daily).

Types

Upsampling: Increase frequency (e.g., monthly → daily). Requires filling missing values.

Downsampling: Decrease frequency (e.g., daily → monthly average).

### Part 2 — Interpolation

Why

Used after upsampling or when data is missing to estimate unknown values.

### Part 3 — Transformation

Why

Transformations reduce variance, make data more linear, and help meet model assumptions.

### Common methods

Square root transform: Reduces spread when variance increases moderately.

Log transform: Handles exponential growth.

### 📌 Stationarity

### Definition

A time series is **stationary** if its **statistical properties** —  
**mean**, **variance**, and **autocorrelation** — remain **constant over time**.

### 📌 Concept

- In **stationary data**, patterns are **stable and predictable** over time.
- Most statistical forecasting methods like **ARIMA (AutoRegressive Integrated Moving Average)** assume the series is **stationary**.
- If a series is **not stationary**, the model may produce **misleading forecasts**.
- Therefore, before building a model, it is important to **check whether the time series is stationary**.  
  If not, **convert it to stationary first**, then build the model.

### 📌 Three Main Types of Stationarity

1. **Strict stationarity** – The **probability distribution** is constant over time.

2. **Weak (or covariance) stationarity** – **Mean**, **variance**, and **autocovariance** do not change over time.

3. **Trend and seasonality effects** – Both break stationarity and need removal before modeling.

### 📌 Why It Matters

If a series has a **trend** or **changing variance**, the model may interpret these  
changes as **permanent rather than temporary**, leading to **poor future predictions**.

### 📌 ADF (Augmented Dickey–Fuller) Test

The ADF test checks if a series is **stationary**:

- **Null hypothesis (H₀):** The series is **non-stationary**.
- **Alternative hypothesis (H₁):** The series is **stationary**.

If **p-value < 0.05**, reject **H₀** → data is **stationary**.


### 📌 ACF and PACF in Time Series Analysis

When working with time series data, it is often useful to analyze the  
**autocorrelation** of the data to understand the **patterns and dependencies**  
between time steps.

Two common tools for this analysis are the **Autocorrelation Function (ACF)**  
and the **Partial Autocorrelation Function (PACF)**.

---

### Definition

**Autocorrelation (ACF):**  
Measures how a time series is related to its **past values** over different **lags**.

ACF plots show the **correlation between a time series and lagged versions of itself**.  
The ACF plot can be used to identify the **number of lags** needed for a time series model.

### 📌 ACF Example

For example, if the **ACF plot** shows a strong correlation between the time series  
and its **lag-1**, **lag-2**, and **lag-3** values, then a time series model that  
includes these lags would likely be a **good fit**.

---

### 📌 PACF — Definition

**Partial Autocorrelation (PACF):**  
Measures the correlation between a time series and its past values **after removing  
the influence of intermediate lags**.

PACF plots show the correlation between a time series and its lagged values **after  
accounting for correlations at shorter lags**.

This is useful for:
- Identifying the **most important lags** to include in a time series model
- Detecting the **presence of seasonal patterns** in the data

### 📌 Concept

- A **lag** is the time difference between observations  
  *(e.g., Lag 1 = previous month, Lag 2 = two months ago)*.

- **ACF plot** → shows correlation for **all lags**, including **indirect effects**.

- **PACF plot** → shows only the **direct effect** of a lag on the current value.

---

### 📌 Purpose in ARIMA

- **PACF** is used to identify the **AR (p)** term.
- **ACF** is used to identify the **MA (q)** term.

### 📌 ACF vs PACF — Intuition

- **ACF** looks at **last week and all the weeks before**  
  *(combined effect)*.

- **PACF** isolates **just last week’s effect**,  
  ignoring the influence of earlier weeks.

### 📌 ACF vs PACF

**ACF (Autocorrelation Function)**  
*"How similar is the series to a shifted copy of itself?"*

- **ACF at lag 1**: correlation between \( x_t \) and \( x_{t-1} \)
- **ACF at lag 2**: correlation between \( x_t \) and \( x_{t-2} \), etc.

ACF includes **direct + indirect effects**  
(e.g., lag 2 includes the pathway through lag 1).

### 📌 PACF (Partial Autocorrelation Function)

*"How similar is the series to a shifted copy **after removing the in-between lags**?"*

- **PACF at lag 2** removes the influence of **lag 1** from both sides,  
  leaving the **direct link** between \( x_t \) and \( x_{t-2} \).


### 📌 Interpreting ACF and PACF Plots

To interpret ACF and PACF plots, we can look for the following patterns:

- A **sharp cutoff in the ACF plot** at a particular lag suggests that a time series model  
  with a corresponding number of lags would be a **good fit**.

- A **slowly decaying ACF plot** may indicate the presence of a **trend** in the data.

- A **slowly decaying PACF plot** may indicate the presence of **seasonality** in the data.

- A **sharp cutoff in the PACF plot** at a particular lag may indicate the presence of a  
  **seasonal pattern** in the data.

> **Rule of thumb:**  
> - Sharp cutoff → specific model order  
> - Slow decay → trend or seasonality

### Quick Recap of ARIMA

ARIMA = Auto-Regressive + Integrated + Moving Average

Works well for stationary series.

Limitation: Cannot naturally handle seasonality.

👉 Example to show: A sales dataset with clear monthly or weekly repeating patterns where ARIMA struggles.


### Why ARIMA Fails on Seasonal Data

ARIMA tries to fit seasonality as noise → results in poor forecasts.

Show ARIMA predictions missing seasonal peaks.

💡 Key Message: ARIMA is powerful, but not seasonal-aware.

### SARIMA as an Extension

SARIMA = Seasonal ARIMA = ARIMA + Seasonal Terms

Formula: SARIMA(p,d,q)(P,D,Q,m)

Non-seasonal part: (p,d,q) → same as ARIMA

Seasonal part: (P,D,Q,m)

P = seasonal autoregression

D = seasonal differencing

Q = seasonal moving average

m = length of season (e.g., 12 for monthly, 7 for weekly)
