# Introduction to NLP (Natural Language Processing)

### What is NLP?
- Natural Language Processing is field of AI that enables computers to understand and process human language.
- It combines linguistics and machine learning to extract meaningful information from text.

### Why is NLP Important?
- We generate huge amounts of text daily, from social media posts and emails to news articles and chat messages. 
- Analyzing such large text data manually is impractical.

#### Key Steps in Text Preprocessing:

1. Tokenization - Breaking text into words or sentences.
2. Stopword Removal - Removing common words that don't add meaning (e.g. - the, is, and).
3. Lemmatization & Stemming - Converting workds to their base form ("running" -> "run")
4. Part-of-Speech Tagging - Identifying grammatical roles (noun, verb, adjective).
5. Named Entity Recognition (NER) - Recognizing proper names, places, organizations.

### 📌 Basic Text Processing Techniques (with Examples)

### 🔹 Tokenization

**What is it?**  
Breaking down text into **tokens** (words, sentences, subwords).

**Why use it?**  
To prepare raw text for analysis — models can’t directly work on **unstructured text**.

**Without Tokenization:**
- Text is treated as one **blob** → no structure or granularity
- Cannot extract meaningful features

### 📌 What are Stop Words?

- Stop words are **common words** in a language that usually **don’t add much meaning**.
- **Examples:** "is", "the", "and", "was", "have", "not"

### 🧠 Why Remove Them?

- They occur **very frequently** in all documents.
- They don’t help in **distinguishing between topics or classes** in many NLP tasks.
- Removing them **reduces noise** and makes models **faster**.

### ⚠️ But Be Careful!

- Not always safe to remove stop words.
- In **sentiment analysis**, removing words like **"not"** can flip the meaning!

### 📌 Stemming

#### What is Stemming?

- A text normalization technique that reduces **inflected or derived words** to their **word stem**.
- It uses **rule-based chopping of suffixes**.

**Example:**
- `"running"` → `"run"`
- `"flies"` → `"fli"`
- `"responsibilities"` → `"respons"`

Popular stemmer in NLTK: Power Stemmer

### 📌 Lemmatization

#### What is Lemmatization?

- A **smarter version of stemming**.
- It reduces words to their **dictionary root (lemma)**, considering the **part of speech (POS)**.

**Examples:**
- `"running"` *(verb)* → `"run"`
- `"better"` *(adjective)* → `"good"`
- `"caring"` → `"care"`



#### When to Use what?

|Scenario|Use|
|-|-|
|Fast and dirty preprocessing|Stemming|
|Precision-focuses applications (QA, Chatbots)| Lemmatization|
|Short text cleanup| Stop word removal|



## Term Frequency - Inverse Document Frequency

#### TF-IDF in NLP

#### What is TF-IDF?
- In NLP, not all words are equally important. Words like 'the', 'is', or 'and' appear in almost every document - they don't help us identify what a specific document is really about. Thats where TF-IDF comes in.

- TF-IDF stands for Term Frequency-Inverse Document Frequency, and its a technique to find out how important a word is in a document compared to a whole collection (called corpus).

#### Definition -
- TF-IDF is statistical measure used to evaluate how important a word is to a document in a collection of documents.

- TF-IDF helps highlight unique and meaningful words while downplaying the common ones.

#### Why do we need TF-IDF?
- One might wonder can't we just count how ofter each word appears in a document and use that as a feature? Thats called Term Frequency (TF).
- But here the problem: some workds occur frequently across all documents like 'the','are' or 'data'.
- They don't tell us anything unique about a specific document. 
- So, just counting words isn't enough. We need a smarter way - and thats where TF-IDF helps!!!



#### How TF-IDF solves it?
- Boost important terms (unique to fewer documents).
- Penalizes common terms (appear everywhere).
- Brings balance between frequency and rarity.


TF-IDF helps focus on words that are both frequent in a document and rare across documents- those are usualy the most meaningful.

### 📌 Alternatives to TF-IDF

| Method                   | Description                                                      | When to Use                                      |
|--------------------------|------------------------------------------------------------------|--------------------------------------------------|
| Bag of Words (BoW)       | Count word occurrences in a document (no weighting).             | Simple tasks, small datasets                     |
| Word Embeddings          | Represent words as vectors using models like Word2Vec, GloVe, FastText | When meaning / semantic similarity matters       |
| TF Only                  | Just counts how often a term appears.                            | Very basic models, or when speed matters         |
| Transformer Embeddings   | Context-aware vectors from models like BERT.                     | Advanced NLP (e.g., summarization, QA)           |


### 📌 TF-IDF vs Alternatives

| Metric              | BoW (Bag of Words) | TF-IDF | Embeddings |
|---------------------|-------------------|--------|------------|
| Word Meaning        | ❌                | ❌     | ✅         |
| Word Weighting      | ❌                | ✅     | ✅         |
| Context Aware       | ❌                | ❌     | ✅         |
| Easy to Implement   | ✅                | ✅     | ❌ (needs training or pre-trained models) |

### 📌 Key Message

TF-IDF is a **sweet spot between simplicity and intelligence** —  
it’s **fast**, **interpretable**, and **powerful** for many real-world tasks.

### 🎓 The Math Behind TF-IDF

- **Term Frequency (TF)** — how often a word appears in a document.
- **Inverse Document Frequency (IDF)** — how rare the word is across documents.

### 1️⃣ Term Frequency (TF)

How frequent is the word in the current document?

$$
\text{TF}(t) =
\frac{\text{Number of times term } t \text{ appears in the document}}
{\text{Total number of terms in the document}}
$$

---

**Example:**

Document:  
> "Cancer is a serious disease. Cancer treatment is expensive."

Term: **"cancer"**

$$
\text{TF} = \frac{2}{8} = 0.25
$$

### 2️⃣ Inverse Document Frequency (IDF)

How rare is the word across all documents?

$$
\text{IDF}(t) = \log\left(
\frac{\text{Total number of documents}}
{\text{Number of documents containing the term } t}
\right)
$$

---

**Example:**

- Total documents = **100**
- `"cancer"` appears in **5** documents

$$
\text{IDF} = \log\left(\frac{100}{5}\right) = \log(20) \approx 1.30
$$


### 3️⃣ TF-IDF

$$
\text{TF-IDF}(t) = \text{TF}(t) \times \text{IDF}(t)
$$

So for the term **"cancer"**:

$$
\text{TF-IDF} = 0.25 \times 1.30 = 0.325
$$


### 📌 Interpretation

- **Higher TF-IDF value** → Word is **important and specific** to the document  
- **Lower TF-IDF value** → Word is **too common** or **not important**
- TF-IDF is just a product of how ofter a term appears and how rare it is across documents.
