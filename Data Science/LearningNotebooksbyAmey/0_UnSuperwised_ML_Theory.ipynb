### 📌 What is Clustering?

- Clustering is a method of organizing data points into **clusters or groups** that share similarities.

- Data points with **similar characteristics** are grouped together,  
  while data points with **dissimilar characteristics** are placed in different clusters.

### 📌 Clustering is Unsupervised ML

- There might be situations where we do **not have any target variable** to predict.

- Such problems, without any fixed target variable, are known as **unsupervised learning problems**.

- In these problems, we only have **independent variables** and **no target/dependent variable**.

- In clustering, we do not have a target to predict.  
  We look at the data and try to **club similar observations** and form different groups.  
  Hence, it is an **unsupervised learning problem**.


#### Properties of Clusters
1. Property 1:
    - All data points in a cluster should be similar to each other.
2. Property 2:
    - The data points from different clusters should be as different as possible.

# K-Means Clustering

##### Step by Step calculation behind K-Means Clustering

1. Select the number K to decide the number of clusters.
2. Select random K points or centroids
3. Assign each data point to their closest centroid, which will form the predefined K clusters.
4. Calculate the variance and place a new centroid of each cluster.
5. Repeat the third step, which means reassign each datapoint to the new closest centroid of each cluster.
6. If any reassignment occurs, then go to step 4 else go to Finish.
7. The model is ready.

### 📌 K-Means Clustering Model Evaluation Technique

#### Definition:
The **silhouette score** is a measure of how similar an object is to its **own cluster** compared to **other clusters**.  
It provides a way to assess the quality of clustering results in **unsupervised learning**.

---

#### Formula:
For each data point **i**:

- **a(i)**: The average distance between point *i* and all other points in the **same cluster**.
- **b(i)**: The minimum average distance between point *i* and all points in **any other cluster** of which *i* is not a member.

The silhouette score **s(i)** for a single data point *i* is calculated as:


$$
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
$$

### 📌 K-Means Clustering Model Evaluation Technique

The **silhouette score** ranges from **-1 to 1**:

- **+1**: Indicates that the sample is **far away from the neighboring clusters** and **well matched** to its own cluster.

- **0**: Indicates that the sample is **on or very close to the decision boundary** between two neighboring clusters.

- **-1**: Indicates that the sample **might have been assigned to the wrong cluster**.


# Hierarchical Clustering -

### Agglomerative with Dendrogram
- Is a method of cluster analysis which seeks to build a hierarchy of clusters.

- A Dendrogram is a tree that shows how clusters are merged step by step.
    - We cut the dendrogram at a certain height to form final clusters.


- Agglomerative Hierarchical clustering starts with each observation as its own cluster and iteratively merges the 2 closest clusters until all points are in a single cluster. The process is visually represented using a dendrogram.

#### Step by Step Process of dendrogram creation -
1. Each data point is treated as its own cluster.
2. Compute pairwise distances between all clusters.
3. Identify the 2 clusters with the smallest distance.
4. Merge the selected pair into a single cluster.
5. Update the distance matrix to include the new cluster.
6. Repeat until all points are merged into one cluster.
7. The dendrogram shows the sequence and distance of merges.
8. A horizontal cut through the dendrogram defines the number of clusters.

#### Linkage Methods

1. Single Linkage (Nearest Nearest Neighbour) - 
    - smallest distance between any points in clusters.
    - Distance between 2 clusters is minimum distancne between any 2 points from the clusters.
2. Complete Linkage - 
    - Farthest distance between any points in cluster.
    - Distance between 2 clusters is the maximum distance between any 2 points from the clusters.
3. Average Linkage - 
    - All distances divided by number of pairs.
    - Distance between 2 clusters is average of all poirwise distances between points in 2 clusters.
4. Centroid Linkage - 
    - Distance between centroid of one cluster and other.



##### Interpreting the Dendrogram
- Leaves represent the original points.
- The height of each node indicates the distance at which clusters were merged.
- Cutting the dendrogram horizontally defines the number of clusters.
- Low merge height means stronger similarity between clusters.

#### Advantages
- Does not require pre - specified of the number of clusters.
- Produces a hierarchy of nested clusters.

#### Limitations 
- Computationally expensive for large datasets.
- Different linkage methods may produce different results.
- Sensitive to noise or outliers.
- Once merged, clusters cannot be split again.

#### Practical Considerations
- Features should be standardized when using Euclidean distance.
- Multiple linkage methods should be tried for comparison.
- Validation can be done using silhouetter score or cophenetic correlation.
