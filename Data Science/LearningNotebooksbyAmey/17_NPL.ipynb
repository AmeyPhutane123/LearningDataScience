```python

# 🎯 Input text
text = "Hello! I'm learning NLP. Let's test tokenization — sentence-wise and word-wise. Amazing, isn't it?"

print("=== 📌 NLTK Sentence Tokenization ===")

sentences = sent_tokenize(text)

# sent_tokenize: Splits text into sentences using NLTK's pre-trained Punkt tokenizer
# - Handles abbreviations, punctuation, and multiple languages
# - Returns list of sentences
```

```python
sentences
```

```python
for i, sent in enumerate(sentences): 
    print(f"Sentence {i+1}: {sent}")


# enumerate adds a counter (i) to each element (sent) in the iterable (sentences), 

# allowing you to loop with both index and value.

# (Example: Turns ['Hello', 'World'] into (0, 'Hello'), (1, 'World').)

    
print("\n=== 🔍 NLTK Word Tokenization ===")

for i, sent in enumerate(sentences):
    print(f"\nSentence {i+1} tokens:")
    words = word_tokenize(sent)
    print(words)
    
    
    
# word_tokenize: Splits text/sentences into words (tokens)
    # - Handles contractions ("I'm" → "I", "'m")
    # - Splits punctuation into separate tokens
    # - Uses TreebankWordTokenizer by default
```

```python
#===============================
# 🚀 spaCy Tokenization Example
# ===============================
import spacy
nlp = spacy.load("en_core_web_sm")  # Load small English model

doc = nlp(text)

print("\n=== 🚀 spaCy Tokenization ===")
print("Tokens:")
print()
print()
for token in doc:
    print(token.text)
print()
print()
print("\n=== ✂️ spaCy Sentence Segmentation ===")

for i, sent in enumerate(doc.sents):
    print(f"Sentence {i+1}: {sent.text}")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')

```

```python
text = "This is not a good product, but it is affordable."
tokens = word_tokenize(text) # Splits the text into individual words/punctuation.

stop_words = set(stopwords.words('english')) #  Loads English stopwords (common words like "is", "a") as a set.

filtered = [word for word in tokens if word.lower() not in stop_words]
# Removes stopwords from tokens, keeping only meaningful words.

print("Without Stop Words:", filtered)
```

```python
from nltk.stem import PorterStemmer

#nltk → The Natural Language Toolkit library for text processing.

#stem → The NLTK submodule containing stemming algorithms.

# PorterStemmer → A class implementing the Porter stemming algorithm (reduces words to root forms, e.g., "running" → "run").
```

```python

stemmer = PorterStemmer()
print(stemmer.stem("running"))         # run
print(stemmer.stem("responsibilities")) # respons
print(stemmer.stem("flies"))            # fli


#Stems "running" → "run" (base form).
```

```python
from nltk.stem import WordNetLemmatizer

#nltk → The Natural Language Toolkit library for text processing.

#stem → The NLTK submodule containing stemming algorithms.

# WordNetLemmatizer ->A class implementing the lemmat algorithm (reduces words to root forms, e.g., "glies" → "fly").
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize("running", pos="v"))  # run
print(lemmatizer.lemmatize("better", pos="a"))   # good
print(lemmatizer.lemmatize("flies",pos='v'))   # fly

# pos="v" → Specifies the part of speech (verb here) for accurate lemmatization (default is noun, "n").
```

```python
# Full example: Stopword removal + Lemmatization
from nltk.tokenize import word_tokenize

from nltk.corpus import stopwords

# Download if not already
nltk.download('stopwords')

# Get English stop words
stop_words = stopwords.words('english')

text = "Cats are running around the house, chasing mice happily."

tokens = word_tokenize(text)
filtered = [word for word in tokens if word.lower() not in stop_words]

lemmatized = [lemmatizer.lemmatize(word.lower(), pos='v') for word in filtered]
print("Processed:", lemmatized)
```
