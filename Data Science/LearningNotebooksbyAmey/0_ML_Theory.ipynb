# Introduction to Supervised Learning

### 1. What is Supervised Learning?
- type of machine learning where a model is trained using labeled data - meaning the input data (features) is paired with the output(target/ label).

    - The goal is for model to learn a mapping function from inputs to outputs so that it can predict the correct label for new, unseen data.
    - The learning process involves adjusting model parameters to minimize prediction errors on the training data.

- Key Components of Supervised Learning: 
    1. Input Features (X): The independent variables used to make predictions.
    2. Outpot Labels (Y): The dependent variable we want to predict.
    3. Traning Data: A dataset with known (X,y) pairs used to teach model.
    4. Model: The algorithm that learns from data.
    5. Learning Algorithm: Adjusts model parameters to minimize the loss (Error).

- Types of Supervised Learning Tasks
    - Supervised learning problems are broadly categorized into 2 types: 
        1. Regression (Continuous Output):
            - Predicts a numerical value. (e.g. - price, sales).
            - Common Algorithms: 
                1. Linear Regression 
                2. Polynomial Regression
                3. Decision Trees (for regression).
        2. Classification (Discrete Output):
            - Predicts a category or a class label. (e.g. - Yes/No, spam/not Spam)
            - Exp - 
                - Email spam detection (binary classification).
                - Handwritten digit recognition (multiclass classification). (e.g. - 0 to 9 - predicting among 10 different categories).
            - Common Algorithms: 
                1. Logistic Regression
                2. Decision Trees.
                3. Support Vector Machines (SVM).
            - Note: Logisctic Regression is a classification algorithm (even though its name has regression).



### 2. Classification Problems

- What is Classification? 
    - is a supervised learning task where the goal is to predict a discrete class label  (category) rathen than a continuous value.
- Types of Classification?
    1. Binary Classification (2 classes) - 
        - Example: 
            - Email: Spam(1) or Not Spam (0).
        - Model Used: 
            - Logistic Regression (predicts probabiltiy of fraud).
    2. Multiclass Classification (> 2 classes) - 
        - Example: 
            - Handwritten Digit Recognition (0-9).
            - Animal Image Classification (Cat, Dog, Bird).
        - Model Used:
            - Multinomial Logistic Regression.
- Why not use Linear Regression for Classification?
    - Issue1 - Output is unbounded (can be <0 or >1), but probabilities must be in [0,1].
    - Issue2 - Linear is sensitive to the outliers ( a single extreme data point can skew predicitons).
    - Solution - Use Logistic Regression, which applies a sigmoid function to map predictions to [0,1].

### 3. Linear Regresion - 
- Supervised learning alrotithm used for predicting a continuous numerical output based on input features. It assumes a linear relationship between input variables (X) and output (Y).

- How it works? 
    1. Training: The model learns coefficients (beeta) that minimize the Mean Squared Error (MSE) between predicted and actual values.
    2. Prediction: Once trained, it predicts y for new inputs.

- Limitations of Linear Regression for Classification 
    - Problem : What if we want to predict a binary outcome.(e.g. - Will student pass or fail?)
        - Linear regression can output values outside [0,1], but probabilities must be between 0 and 1.
        - Example - if y = 5, how do we interpred this as a probability?
    - Solution: 
        - We need a function to constrain predictions between 0 and 1 -> Logistic Regression.
        



##### Logistic Regression: The Concept From Linear to Logistic Regression

- Linear Regression: Predicts continuous y using a straight Line.
- Logistic Regression: Predicts probability P(y =1|x) ( Probability of Y = success when x is given) using a S - shaped Curve (sigmoid).
- Sigmoid FUnction (Logistic Function):
    - Output always between 0 and 1.
    - At z=0, P(y=1) = o.5.
    - Approaches 1 as Z goes to positive Infinity.
    - Approaches 0 as Z goes to negative Infinity.

##### Training Logistic Regression (Maximum Likelihood Estimation - MLE)
- Unlike linear Regression (which uses Least Squares), Logistic Regression uses MLE to find coefficients.
- Goal:
    - Choose Beeta values that maximize the likelihood of observing the actual data.


##### Model Evaluation Metrics:
- Since accuracy alone can be misleading
    - Confusion Matrix:
        1. True Negative - when Actual is 0 and Predicted is 0.
        2. True Positive - when Actual is 1 and Predicted is 1.
        3. False Positive - when Actual is 0 and Predicted is 1.
        4. False Negative - when Actual is 1 and Predicted is 0.
        - Simple 
            - If matches - True else False.
            - If Predicted is 0 - its Negative, 1 its Positive.
    - Metrics: to evaluate models
        - Accuracy = (TP+TN)/(TP+TN+FP+FN).
        - Precesion = TP/(TP+FP) (How many predicted positives are correct)?
        - Recall = TP/(TP+FN) (How many actual positives were caught?)
        - F1-Score = Harmonic meann of precision and recall.
        - ROC-AUC = Measures models ability to distinguish classes.


##### Assumptions and Limitations about Logistic Regression: 
-   
    1. Linear Decision Boundary: Assumes as linear relationship between features and log-odds.

    3. Large Sample Size: Words best with sufficient data.

### 4. Decision Trees

- Divide and Conquer
- Decision trees are built using a heuristic called recursive partitioning.
- This approach is also commonly known as divide and conquer because it splits the data into subsets, which are then split repeatedly into even smaller subsets, and so on and so forth until the process stops when the algorithm determines the data within the subsets are sufficiently homogenous (i.e not further division can be done), or another stopping criterion has been met.
- TO see how splitting a dataset can crease a decision tree, imagine a bare root node that will grow into a mature tree. At first, the root node represents the entire dataset, since no splitting has transpired.
- Next, the decision tree also must choose a feature to split upon; ideally, it chooses the feature most predictive of the target class.
- The examples are then partitioned into groups according to the distinct values of this feature, and the first set of tree branches are formed.

- The decision tree hopes to find splits that reduce entropy, ultimately increasing homogeneity within the groups.

### CART (Classification and Regresstion Trees)

- stands for Classification and Regression Trees.
- The beauty of CART is that it worrks for both: 
    1. Classification Problems (predicting categories)
    2. Regression Problems (predicting continuous values)
- think of CART like a flowchart-like structure where: 
    - each internal node represents a decision on a feature.
    - Each leaf node represents the final prediction.

##### CART for Classification: Enter Gini Impurity

- When we use CART for classification (e.g. - smam vs not spam), we need a way to measure how pure or impure a node is. 
- Thats where Gini Impurity comes in
- What is Gini Impurity?
    - Imagine you are blindfolded and picking a ball from a bag.
    - If bag has balls of only one color: you are 100% correct every time -> Gini impurity = 0 (pure node)
    - If bag has mixed colors, you have a chance of being wrong -> Gini Impurity >0.
- How CART Uses Gini Impurity
    - CART looks for the split that reduces Gini Impurity the most.
    - Each ppossible split is like a potential question we could ask.
    - The goal? Find the question that makes the child nodes purer than the parent.
    - "Remember: lower Gini -> better split!"

##### CART for Regression: Enter Least Squares

- In regression (predicting numbers), we can't use Gini.
- Instead, CART uses the idea of Residual Sum of Squares (RSS).
- Concept: In each node, we compute the mean value of the target.
- When then measure the squared difference between the actual values and this mean ( the residuals).
- CART looks for splits that minimize these residuals..
- In Simple Words: 
    - In regresion, the tree is trying to keep similar numbers together- so the average of the group is close to all the values inside it.

##### Limitations of Decision Trees
-   
    1. Overfitting: 
        - Decision Trees can become too complex, capturing noise instead of the true pattern.
        - A tree that's too deep may work perfectly on training data but perform poorly on new, unseen data.
    2. Instability ( High Variance): 
        - Small changes in data can result in completely different trees.
        - This makes decision trees sensitive and less robust.
    3. Bias toward Features with More Levels:
        - Decision Trees can favor features with may unique values (e.g. - ID numbers), even if they are not meaningful

##### Pruning: The Solution to Overfitting 
-   - means cutting back the tree to remove parts that don't add much predictive value.
    - 2 Types :
        - Pre-Pruning (Early Stopping):
            - Stop the tree from growing beyond a certian depth or when the numberr of samples in a node is small.

        - Post Prunin (Reduced Error or Cost Complexity):
            - First grow the full tree, then remove branches that don't significantly improve performace.

#### Problems with Single Model:

So how about instead of trusting one model, we trust mulltiple? This is the core idea behind Ensemble Learning.

#### Definition
Ensemble Learning is a machine learning technique where multiple models (often called base learners or weak learners) are trained and their outputs are combined to produce a stronger, more accurate predictin.

#### Why Ensemble Learning Works
- Reduces Variance: Combats overfitting by averaging predictions (Bagging).
- Reduces Bias: Boosting helps correct weak learners.
- Improves Accuracy: Better than most individual models.
- Stable and Robust: Works well with different types of data.

#### Types of Ensemble Learning Methods:
1. Bagging - 
    - Strategy: Train on bottstrapped samples.
    - Combines by: Average (Regression Problem - Continuous Values)/Vote (Classification Problem - majority winds).
    - Key Use Case: Reduce Variance.
2. Boosting - 
    - Strategy - Sequentially correct mistake
    - Combines By: Weighted Vote
    - Key Use Case: Improve generalization
3. Stacking - 
    - Strategy: Train multiple models + a meta-model
    - Combines By: Meta-model
    - Key Use Case: Improve generalization

#### Fondations:

1. Bagging - Bootstrap Aggergating ( Foundation of Random Forest):
    - Definition: 
        - Ensemble technique where multiple versions of a model are trained on different bootstrapped samples of the dataset and combined via averaginng (regression) or voting (classification).
    - Process: 
        - Bootstrap Sampling: Draw random samples with replacement from training data.

        - Aggreegate Results:
            - For Classification: Majority Voting

#### What is Random Forest:

- Concept and Definiton: Ensemble Machine Learning Algorithm that constructs a large number of decision trees at training time and outputs the model (classification) or mean (regression) of the individual trees.
- It improves prediction accuracy by reducing variance through:
    - Bootstrap sampling.
    - Random feature selection at each split.

#### Why "Random" + "Forest"?
- Forest - Collection of decision trees
- Random - rondomness in both:
    - Row Sampling (Bagging)
    - Column Sampling (Random subset of features)

- Goal: Make each tree different: Reduce Correlation -> Reduce Overfitting.

#### How Random Forest Works - Step by Step

1. Bootstrap Sampling: Randomly select data with replacement to train each tree.
2. Feature Subsampling: At each node split, choose a random subset of features.
3. Train Decision Tree: Groow each tree fully (or with llimits like max_depth).
4. Prediction Aggregation:
    - Classification : Majority vote
    - Regression : Average of outputs

#### Maths behind Random Forest

The Problem wih Single Decision Tree
- Decision Trees are:
    - <span style="color:green;">Easy to Interpret</span>
    - <span style="color:green;">Fast to Train</span>
- But:
    - **They overfit** (especially when deep)
    - **Very Sensitive** to nice (high variance)
    - Give **Unstable Predictions** if data changes.
- So we need something more **Stable and Generalizable**.

#### The Idea of Random Forest
- What if we create many decision trees, each a bit different, and combine their predictions?
    - Each tree might make mistakes..
    - But together, they average out those errors.
    - This makes the overall precition **more accurate and stable**.
- Thats Rando Forest:
    - Random trees (thanks to data and feature samplling).
    - Forest - a lot of trees working together.

#### Bootstrap Sampling (Data Randomness) :
- Instead of using all training data to trian each tree:
    - We take a random sample with replacement - called bootstrap sample.
    - This created different datasets for different trees -> now each tree will learn slightly differnt rules.

- At every split in a tree, we don't consider all features.
- Instead:
    - We randomly pick a few features (exp - pick 3 out of 10).
    - Then find the best split among those.
- Why?
    - Prevents dominant features from always controlling splits
    - Forces trees to be more diverse.
    - Reduces **correlation** among trees - this is key.

#### Variance Reduction - The Maths insight
- VAriance - how much your prediction fluctuates if training data changes.
    - A deep decision tree has high variance.
    - But if we average many such models - the result becomes stable.

#### Now the Math Intuition
- Imagine:
    - 1 model makes an error of +- 10%
    - But if you have 10 independent models, and average them...
    - Their erros cancel other out - net error is much smaller.

#### Random Forest: 
    - Uses many trees.
    - Each Trained on a different data + features
    - So their predictions are not identical, and not fully correlated.

#### ✏️ Actual Variance **Reduction** Formula

- Each tree has variance = **σ²**
- Trees are correlated: average correlation = **ρ (rho)**
- You use **T** trees

Then total ensemble variance:

$$
\mathrm{Var}_{\text{forest}} = \rho \sigma^2 + \frac{1 - \rho}{T}\sigma^2
$$

#### What this means: 
|Factor|Effect of Variance|
|---|---|
|More Trees (T)|Decreases Variance|
|Lower Correlation ($\rho$)|Decreases Variance|
|Lower Base Variance ($\sigma^2$)|Helps, but not always possible|

#### Summary

|Step|What Happens|Why it happens|
|-|-|-|
|1|Bootstrap sample (row sampling)|Makes each tree different|
|2|Random Feature Selection|Foreces diverse decision paths|
|3|Train many trees|Builds many week learners|
|4|Aggregate predictions (vote/average)|Reduces variance, improves accuracy|
|5|Final Result|A strong, stable predictor|

#### Key Characteristics That Suit :

## Boosting - AdaBoost Machine Learning Algorithm

#### Boosting: Step by Step

##### Definition:
- Boosting is an ensemble technique that builds models sequentially, where each new model focuses on correcting the errors made by the previous ones.

- Core Idea:
    - Models are trained one after the other.
    - Each model learns from the mistakes of the previous.
- Final prediction is a weighted vote (classification) or weighted average (regression).

##### Characteristics of Boosting
|Feature|Boosting|
|-|-|
|Model Sequence|Trained Sequentially|
|Focus| Reduce Bias|
|Feature|Boosting|
|Base Learners|Weak (e.g. - shallow trees)|
|Output Combination|Weighted Vote/ average|
|Overfitting|Can overfit if not regularized|
|Performace|Generally better accuracy than bagging|

##### Why Use Boosting?
- Captures complex patterns.
- Ofter outperforms single models and even bagging.
- Great for imbalanced classification, credit scoring, fraud detection, etc.

##### AdaBoost: Intuition

- How it works:
    - Start with a base model (usually a shallow decision Tree).
    - Give equal weight to all training samples.
    - After each iteration:
        - Increase the weight of **misclassified** samples.
        - Decrease the weight of **correctly classified** ones.
    - Train next model on the updated weights.
    - Final prediction: Weighted vote of all models.
    - Repeat for n estimations

#### 📐 Basic Math

At each iteration **m**:

- Misclassified samples get higher weight.
- Each model gets a score  
  $$
  \alpha_m = \ln\left(\frac{1 - \text{error}_m}{\text{error}_m}\right)
  $$
- Final prediction:  
  $$
  \text{Final prediction} = \mathrm{sign}\left(\sum \alpha_m \times h_m(x)\right)
  $$

**Where:**

- $\text{error}_m$ is the error rate of model **m** — how often it misclassifies points.
- $\alpha_m = \ln\left(\frac{1 - \text{error}_m}{\text{error}_m}\right)$ gives more weight to better models  
  (lower error ⇒ higher $\alpha_m$).
- $h_m(x)$ is the prediction of the **m-th weak learner** for input **x**.
- Final prediction uses a **weighted vote** of all weak learners.





# Boosting - (Gradient Boosting ML Algorithm)

- A powerful boosting technique where each new model tries to correct the errors of the previous one - using gradients of the loss function to guide it.

- Analogy - Imagine you are guessing someone's weight. your first guess is 65kg. They say "Too Low!" So you adjust to 70kg. Still low? You go to 75kg..
- This is how **gradient boosting** works - step by step correction toward the right answer, minimizing the loss (error) each time.

##### How it works: 
1. Start with a weak model (shallow tree).
2. Compute Error (residuals or gradients).
3. Fit a new model to predict the rounds.
4. Repeat for multiple rounds.
5. Final Prediction = Sum of all Small corrections.

### 📌 Key Differences from AdaBoost

| AdaBoost | Gradient Boosting |
|---------|-------------------|
| Adjusts sample weights | Fits to residuals (gradients) |
| Focus on misclassified points | Focus on minimizing <b>loss function</b> |
| Not easily extensible to custom loss | <b><span style="color:green;">Flexible: supports regression, classification, ranking</span></b> |
| Less robust to overfitting | Better regularization options |

### 📌 Mathematical Intuition

Let’s say we are minimizing **log-loss**.

Gradient Boosting builds models that predict **how much to change the output to reduce this loss**, step-by-step.

- Loss: $L(y, \hat{y})$
- Compute gradient: $\frac{dL}{d\hat{y}}$
- New model predicts this gradient
- Update prediction:  
  $$
  \hat{y}_{\text{new}} = \hat{y}_{\text{old}} - \text{learning\_rate} \times \text{gradient}
  $$
